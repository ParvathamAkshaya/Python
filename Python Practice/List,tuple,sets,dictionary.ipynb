DATA STRUCTURE:
It is used to organize and store data that can be easily accessed, managed and modified.
1.List
2.Tuple
3.Set
4.Dictionary

List:It is used to store multiple items in a single variable.

1.It is represented in[]
2.Insertion order is preserved in list.
3.Heterogenous elements are allowed.(all kind of data can take in single element)
4.Duplicates values are also allowed.
5.It is mutable(can be changed)

```python
list=["apple","banana","grapes","grapes"]
print(list)
print(len(list))
```

```python
l1=["app","dapp"]
l2=[1,2,3]
l3=[True, False, False]
l4=["acs",2,True]
print(l1,l2,l3,l4)
print(type(l1))
```

```python
List=["apple","banana","guava"]
print(List[1])
print(List[1:3])
```

```python
#check if item exists
List=["apple","banana","guava"]
if "apple" in List:
    print("yes, 'apple' in list")
```

```python
#Change item list
List=["apple","banana","guava"]#0,1,2
List[1]= "orange"
print(List)
```

```python
#insert() without replacing any existing value.
List=["apple","banana","guava"]
List.insert(2,"orange")
print(List)
```

```python
#append() To add an item at the end of list.
List=["apple","banana","guava"]
List.append("pome")
print(List)
```

```python
#extend() To append elements from another list to current list.
List1=["apple","banana","guava"]
List2=["apple","banana","guava"]
List1.extend(List2)
print(List1)
```

```python
List=["apple","banana","guava"]
List.remove("apple")
print(List)
```

```python
List=["apple","banana","guava"]
List.pop(2)
print(List)
```

```python
#Slicing: Taking a part or subset of a list.
l1=[10,20,30,40,50]#01234..-5-4-3-2-1
print(l1[:3])
print(l1[2:])
print(l1[:])
print(l1[::2])
print(l1[::-1])
print(l1[::-2])
print(l1[-4:-2])
print(l1[-2:-4])
```

TUPLE:
1. It is represented in'()'.
2. Insertion order is preserved in tuple.
3. Heterogenous elements are allowed.
4. duplicate values are allowed.
5. It is immutable(that cannot be changed)
6. it is also called as READ ONLY VERSION OF LIST.

```python
#creation of tuple
t1=(11,)
print(t1)
print(type(t1))
t2= 10,2.3,True,10,'alla',20
print(t2)
print(type(t2))
```

note: If we want to append anything it get error becoz tuple is immutable.
we cannot add or remove but rest of properties as SET we can perform.

SET

SET is known for UNIQUENESS
1. It is represented in '{}'.
2. Insersion order is not possibe.
3. Heterogenous elements are allowed.
4. Duplicates values are not allowed.
5. It is mutable(That can change)

```python
#creation of set
s1={10}
print(s1)
print(type(s1))
s2={10,20,30,10,40,20,20.3,True,"Akki"}# not print duplicates
print(s2)
print(type(s2))
```

```python
#to add elements add(), update()
s3={10,20}
s3.add(30)
print(s3)
S3={44,444,55,66}
s3.update([34,43,99])#list,tuple,string,range(iterable)=eg:index by index
print(s3)
s3.update('python')
print(s3)
```

```python
s4={10,20,30,40,50,60}
s4.remove(10)
print(s4)
s4.pop()
print(s4)
s4.discard(50)
print(s4)
```

DICTIONARY

1. It is represented in key:value term'{k:v}'.
2. Insersion order is not possibe.
3. Heterogenous elements are allowed.
4. Duplicates values are allowed but KEYS must be unique.
5. It is mutable(That can change)

```python
#creation of dictionary
d= {1:"python",'b':200,3:200}
print(d)
print(type(d))
```

```python
# how to update
d= {1:"a",2:"b"}
d[3]="c"
print(d)
```

```python
del d[3]
print(d)
```

```python
d.clear()
print(d)
```

```python
d= {1:"a",2:"b"}
print(d)
print(d.keys())
print(d.values())
print(d.items())
```
