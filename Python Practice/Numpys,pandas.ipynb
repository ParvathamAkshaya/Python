NumPy:Numerical python it is a kind of librabry which works with array.
Array:It is a collection of homogenous element(same data type).
- stored in contiguous memory.
- we can use NumPy while working with large datatypes,numbers,cal,matrics,statistics(sum,avg,mul,matrix oper)
- import numpy as np
- even if we use list y numpy means:Fast,consume less memory,easy for cal
- vector: it is simply a list/array of numbers arranged in one direction.

```python
import numpy as np
arr=np.array([1,2,3,4,5])#row vector
print(arr)
```

```python
#addn operation
a=np.array([1,3,4,5,2,5,6])
print(a+3)
print(a*3)
print(a++3)
print(a-3)
```

```python
#n.dim: it will give the dimension of array.
a.ndim
```

```python
#n.shape:It is a tuple which returns the no.of elements per axis
a.shape
```

```python
#n.dtype: It returns data type
a.dtype
```

```python
#n.size: It counts the no.of elements
a.size
```

```python
data=np.array([[1,2,3],[3,55,67],[00,00,000]])
print(data)
```

```python
data.size
```

```python
data.shape#no.of rows and columns
```

```python
d1=data[:,0:2]#from data 2 columns
d1
```

```python
#statistical operation
d2=data[:,0:1]
d2
```

- mean:addn of all values/no.of values
- meadian:exact middle point after sorting(0,1,3)
- mode:Frequent occurance of any value
- standard deviation:deviation from mean
- var:spread of data

```python
print ("Mean:",np.mean(d1))
print ("Median:",np.median(d1))
#np has no attribute mode
```

```python
print ("Min:",np.mean(d2))
print ("Max:",np.mean(d2))
```

```python
print ("Standard deviation:",np.mean(d1))
print ("Variance:",np.mean(d1))
```
