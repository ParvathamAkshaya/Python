LAMBDA FUNCTION
1. It is a small anonymous function(a function without a name)
2. It is mainly used when you need a simple function for a short period of time.
3. Syntax: lambda arguements: expression
4. arguments:input values
5. expression:single line logic/return value
6. mostly used with map(),filter(),sorted().

```python
cube=lambda x : x**3#Lambda function
cube(3)
```

```python
add=lambda x,y:x+y
add(10,20)
```

```python
p=add(10,20)
print(p)
```

```python
#map()
nums=[1,2,3,4,5]
result=list(map(lambda x:x * 2, nums))
print(result)
```

```python
#filter()
nums=[1,2,3,4,5,6]
even=list(filter(lambda x:x % 2 ==0, nums))
print(even)
```

```python
#sort()
a=[23,45,1,2,3,6,8]
a.sort(key=lambda x:x)
print(a)
```

```python
#sort string
a=["akki","sravs","deepu"]
a.sort(key=lambda x:x)
print(a)
```

```python
employees =[("a",5000),("b",8000),("c",7000)]
employees.sort(key=lambda x:x[1])
print(employees)
```

main imp points
1. Sort()=change original list
2. sorted()=return new list
3. default=asc, reverse =true==descending
4. key=lambda==custom sorting
5. key: It tell python what value it should look at while sorting
