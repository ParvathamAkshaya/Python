REGEX(Regular expression)

1. A RegEx is a sequence of characters that form a search pattern.
2. it can be used to check if a string contains the specified search pattern.
3. Regex module:
4. python has a built-in package called 're' while can be used to work with regular expression.
5. 
   REDEX FUNCTIONS
   
- findall: returns a list containing all matches.
- search: returns a match object if there is a match any where in the string.
- split: returns a list where the string has been split each match
- sub: replaces one or many matches with a string

Meta characters: these are characters with special meaning.

```python
import re
txt= "Hello everyone its rain raining"
x= re.findall("rai",txt)
print(x)
```

```python
txt= "Hello everyone its rain raining"
x= re.findall("por",txt)
print(x)
```

```python
str1='the match2 is going on in spain, match2'
re.findall("[aeiou]",str1)#search for all char[aeiou] in string
```

```python
str1='the match2 is going on in spain, match2'
print(re.findall("[^aeiou]",str1))#returns values other than the given
```

```python
str1='the match2 is going on in spain, match2'
re.findall("^[t]",str1)#check weather string start with "t"
```

```python
text="hello planet, welcome, helooooo,hi"
re.findall("^hello",text)#check weather string start with hello
```

```python
text="hello planet welcome"
re.findall("he..o",text)
```

```python
text="helllo planet welcome"
re.findall("he...o",text)#find exact 3 letters in b/w h & o
```

```python
text="hello planet, welcome, helooooo,hi"
re.findall("he.*o",text)
```

```python
text="hello planet, welcome, helooooo,hi"
re.findall("hi$",text)#check sring end with "hi"
```

```python
text="hello planet, welcome, helooooo,hi"
re.findall("i$",text)#check sring ends with "i"
```

```python
text='Hi welcome to python2,285'
re.findall('\d',text)
```

SEARCH() FUCTION:
It searches the string for a match and returns a match object 
if there is more than one match only first occurance of match will be returned.

```python
text="The rain in spain"
x= re.search("\s",txt)
print("The first white space position:",x.start())
```

1. span() returns a tuple containing the start-, and end positions of the match.
2. string returns the string passed into the function
3. group() returns the part of the string where there was a match

- \b==word boundary(start of a word)
- "S"UPPERCASE
- \w+== one or more word characters

```python
import re
txt="The rain in spain"
x= re.search(r"\bs\w+",txt)
print(x)
```

GROUP
- Print the part of the string where there was a match.
- the regex looks for any words that start with UPPERCASE"S"

```python
import re
txt="The rain in spain"
x= re.search(r"\bs\w+",txt)
print(x.group())
```
