MATCH: It is the statement used to perform different actions based on different conditions.
HOW IT WORKS:
1. The match expression is evaluated once.
2. The value of expression is compared with values in each case.
3. if there is a match the associated block of code is executed.

```python
day=3
match day:
    case 1:
        print("monday")
    case 2:
        print("tuesday")
    case 3:
        print("wed")
    case _:
        print("no day")
        
```
