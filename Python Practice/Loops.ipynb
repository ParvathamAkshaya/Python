```python
LOOPS
Loops: Repeating block of code multiple times.
python has two primitive loops:
1. While loop
2. For loop
```

```python
#while loop
i=1
while i<6:
    print(i)
    i+=1
```

```python
i=1
while i<=6:
    print(i)
    if i == 3:
        break
    i+=1
```

```python
i=10
while i>=1:
    print(i)
    i-=1
```

```python
i=0
while i<6:
    i+=1
    if i ==3:
        continue
    print(i)
```

```python
#else statement
i=1
while i<6:
    print(i)
    i+=1
else:
    print("i is no longer less than 6")
```

```python
#for loop
#A for loop is used for iterating over a sequence ( that is either a list, a tuple,a dictionary, a set, or a string)
fruits = ["apple", "banana", "cherry"]
for x in fruits:
    print(x)
```

```python
for x in "banana":
    print(x)
```

```python
for x in range(7):
    print(x)
```

```python
for x in range(2,7):
    print(x)
```

```python
for x in range(2,30,3):
    print(x)
```

```python
for x in range(7):
    print(x)
else:
    print("finally finished")
```

```python
for x in range(7):
    if x == 3:break
    print(x)
else:
    print("finally finished")
```

```python
#nested loop: it is a loop inside a loop
cars=["toyota" , "benx"]
fruits=["apple" , "banana"]
for x in cars:
    for y in fruits:
        print(x,y)
```

```python
#pass statement: loops cannot be empty,if loop with no content just put pass for no error.
for x in [0,1,2]:
    pass

```
