```python
#How to access element?
#forward index:0 1 2 3 4 5  Reverse index: -6 -5 -4 -3 -2 -1
s='python'
#index
print(s[-2])
print(s[-1])
print(s[2])
print(s[4])
```

```python
s='python'
#slicing:Take a part of list
print(s[::])
print(s[:])
print(s[:2])#py
print(s[4:])#on
print(s[::-4])#ny
print(s[::-3])#nt
print(s[::-1])#nohtyp
print(s[-3:-5:-1])#ht
print(s[-2:-5:-1])#oht
```

```python
s= "  String it is a kind of datatype  "
print(s.strip())
print("lstrip:",s.lstrip())#remove space from left end
print("rstrip:",s.rstrip())#remove space from right side
```

```python
s= "python programming is amazing"
#index(),find():both return index position.....find is best becoz it not throw error
#find():Returns index position of the value.if value is not found it return -1 but index() it throw error if value is not found
print(s.index('p'))
print(s.index('python'))
print(s.index('programming'))
print(s.index('is'))
print(s.index('y'))
print(s.find('y'))
print(s.find('p',1,8))
print(s.find('b'))#no value but it return -1
```
