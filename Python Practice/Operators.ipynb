```python
Arithematic operators: An operator which is used to perform mathematical operations on numeric values.
(+,-,*,/,%,**,//)
```

```python
x=15
y=5
print(x+y)#Addition
print(x-y)#subtraction
print(x*y)#multiplication
print(x/y)#division
print(x%y)#modulus(remainder)
print(x**y)#exponentitation(power)2**3=2*2=4*2=8//8 is answer
print(x//y)#floor division(no decimal)
```

Assignment operators:An operator which is used to assign values to variables.
(=,+=,-=,*=,/=,%=,//=,**=,&=,|=,^=,>>=,<<=,:=)

```python
x=5
x+=3
print(x)
```

```python
x=5
x-=3
print(x)
```

```python
x=5
x%=3
print(x)
```

```python
x=5
y=3
x+=y
print(x)
x-=y
print(x)
x*=y
print(x)
x/=y
print(x)
x%=y
print(x)
x**=y
print(x)
x//=y
print(x)
```

```python
#bitwise assignment
x=5
y=3
x&=y
print(x)
x|=y
print(x)
x^=y
print(x)
x<<=y
print(x)
x>>=y
print(x)
```

Comparision operator:It is used to compare two values and return true or false
(==,!=,<,>,<=,>=)

```python
x=6
y=8
print(x==y)
print(x!=y)
print(x<y)
print(x>y)
print(x<=y)
print(x>=y)
```

```python
#Chaining comparision
x=5
print(1<x<10)
print(3<x and x >10)
```

Logical operator: It is used to combine conditional statement.
(and,or,not)

```python
x=5
y=10
print(x>0 and x<11)
print(y>10 or y<11)
print(not(x>3 and x<10))
```

Identity operators:It is used to compare the objects.
(is,isnot)

```python
x=["apple","banana"]
y=["apple","banana"]
z=x
print(x is z)
print(x is y)
print(y is not z)
print (x==y)
```

Membership operators: It is used to test if a sequence is presented in an object.
(in,not in)

```python
cars=["bmw","suzuki","benz"]
text= "hello"
print("bmw" in cars)
print("benz" not in cars)
print("gvvhjg"not in cars)
print("h" in text)
```
